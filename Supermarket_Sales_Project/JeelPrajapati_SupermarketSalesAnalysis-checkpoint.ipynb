# Supermarket Sales Analysis

## Data Analytics Project

### Objective
The aim of this project is to analyze supermarket sales data and find useful
information about products, branches, categories, customers, payment methods,
and customer ratings.

### Dataset
The dataset contains 500 supermarket sales transactions.

# Step 1 — Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

# Step 2 — Load Dataset


```python
df = pd.read_csv("SUPER MARKET DATA - supermarket_sales_500_rows.csv")

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df.tail()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

### Main Columns

- Invoice ID – Unique invoice number
- Date – Date of transaction
- Branch – Supermarket branch
- City – City of the branch
- Customer Type – Member or Normal customer
- Gender – Customer gender
- Product – Product purchased
- Category – Product category
- Quantity – Number of products purchased
- Unit Price – Price of one product
- Payment – Payment method
- Rating – Customer rating
- Sales – Total sales amount

```python
df.info()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

# Step 3 — Check Duplicate Rows

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

# Step 4 — Convert Date

```python
df["Date"] = pd.to_datetime(df["Date"])

print("Date converted successfully.")
print(df["Date"].dtype)
```

# Step 5 — Check Incorrect Values

```python
print("Quantity minimum:", df["Quantity"].min())
print("Unit Price minimum:", df["Unit Price"].min())
print("Rating minimum:", df["Rating"].min())
print("Rating maximum:", df["Rating"].max())
```

```python
print("Unique Customer Types:")
print(df["Customer Type"].unique())

print("\nUnique Branches:")
print(df["Branch"].unique())

print("\nUnique Categories:")
print(df["Category"].unique())

print("\nUnique Payment Methods:")
print(df["Payment"].unique())
```

# Basic Statistics

```python
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
total_quantity = df["Quantity"].sum()
average_rating = df["Rating"].mean()
total_transactions = df["Invoice ID"].nunique()

print("Total Sales: ₹", round(total_sales, 2))
print("Average Transaction Sales: ₹", round(average_sales, 2))
print("Total Quantity Sold:", total_quantity)
print("Average Customer Rating:", round(average_rating, 2))
print("Total Transactions:", total_transactions)
```

# Product Analysis

```python
product_sales = (
    df.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(product_sales)
```

```python
highest_product = product_sales.idxmax()
highest_product_sales = product_sales.max()

print("Highest Sales Product:", highest_product)
print("Sales:", round(highest_product_sales, 2))
```

# Product Sales Chart

```python
plt.figure(figsize=(10, 5))

product_sales.sort_values().plot(kind="barh")

plt.title("Sales by Product")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

# Top 5 Products

```python
top_5_products = product_sales.head(5)

print("Top 5 Products by Sales:")
print(top_5_products)
```

```python
plt.figure(figsize=(8, 5))

top_5_products.plot(kind="bar")

plt.title("Top 5 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Branch Analysis

```python
branch_sales = (
    df.groupby(["Branch", "City"])["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(branch_sales)
```

```python
best_branch = branch_sales.idxmax()
best_branch_sales = branch_sales.max()

print("Best Performing Branch:", best_branch)
print("Sales:", round(best_branch_sales, 2))
```

```python
branch_only_sales = df.groupby("Branch")["Sales"].sum()

plt.figure(figsize=(7, 5))

branch_only_sales.plot(kind="bar")

plt.title("Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Category Analysis

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

print("Highest Selling Category:", highest_category)
print("Sales:", round(highest_category_sales, 2))
```

# Payment Method Analysis

```python
payment_count = df["Payment"].value_counts()

print(payment_count)
```

```python
most_used_payment = payment_count.idxmax()
most_used_payment_count = payment_count.max()

print("Most Used Payment Method:", most_used_payment)
print("Number of Transactions:", most_used_payment_count)
```

```python
plt.figure(figsize=(7, 5))

payment_count.plot(kind="bar")

plt.title("Payment Method Usage")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Member vs Normal Customers

```python
customer_type_sales = (
    df.groupby("Customer Type")["Sales"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Sales by Customer Type:")
print(customer_type_sales)
```

```python
for customer_type, amount in customer_type_sales.items():
    print(customer_type, ": ₹", round(amount, 2))
```

```python
plt.figure(figsize=(7, 5))

customer_type_sales.plot(kind="bar")

plt.title("Average Spending: Member vs Normal")
plt.xlabel("Customer Type")
plt.ylabel("Average Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Customer Rating Analysis

```python
average_rating = df["Rating"].mean()

print("Average Customer Rating:",
      round(average_rating, 2),
      "out of 5")
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Rating"], bins=10)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

# Gender Analysis

```python
gender_sales = df.groupby("Gender")["Sales"].sum()

print(gender_sales)
```

```python
plt.figure(figsize=(7, 5))

gender_sales.plot(kind="bar")

plt.title("Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Quantity Analysis

```python
product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print(product_quantity)
```

```python
most_sold_product = product_quantity.idxmax()
most_sold_quantity = product_quantity.max()

print("Most Sold Product:", most_sold_product)
print("Total Quantity Sold:", most_sold_quantity)
```

# Monthly Sales Analysis

```python
df["Month"] = df["Date"].dt.month_name()

monthly_sales = (
    df.groupby("Month")["Sales"]
    .sum()
)

print(monthly_sales)
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_sales = monthly_sales.reindex(
    [month for month in month_order if month in monthly_sales.index]
)

print(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

monthly_sales.plot(kind="bar")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Sales Summary Table


```python
summary = pd.DataFrame({
    "Metric": [
        "Total Transactions",
        "Total Sales",
        "Average Transaction Sales",
        "Total Quantity Sold",
        "Average Customer Rating"
    ],
    "Value": [
        total_transactions,
        round(total_sales, 2),
        round(average_sales, 2),
        total_quantity,
        round(average_rating, 2)
    ]
})

summary
```
